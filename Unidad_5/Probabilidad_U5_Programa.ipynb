```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import statsmodels.api as sm

# Lista de archivos a analizar
archivos = ['grupo1.csv', 'grupo2.csv', 'grupo3.csv']

for archivo in archivos:
    print(f"\n==============================================")
    print(f"ANÁLISIS PARA: {archivo}")
    print(f"==============================================")

    # 1. Carga y exploración de datos
    df = pd.read_csv(archivo)
    print("\n--- Primeras filas ---")
    print(df.head())
    print("\n--- Estadística descriptiva ---")
    print(df.describe())

    x = df.iloc[:, 0] # Horas de Estudio
    y = df.iloc[:, 1] # Calificación Final

    # 2. Coeficiente de correlación de Pearson
    r, p_value_corr = stats.pearsonr(x, y)
    print(f"\nCoeficiente de Correlación de Pearson (r): {r:.4f}")
    print(f"Valor p de correlación: {p_value_corr:.4e}")

    # 3. Modelo de Regresión Lineal (Statsmodels)
    X_sm = sm.add_constant(x)
    modelo = sm.OLS(y, X_sm).fit()
    beta_0 = modelo.params[0]
    beta_1 = modelo.params[1]
    r_squared = modelo.rsquared
    p_value_slope = modelo.pvalues[1]

    print(f"\nEcuación de Regresión: Y = {beta_0:.4f} + {beta_1:.4f} * X")
    print(f"Coeficiente de Determinación (R^2): {r_squared:.4f}")
    print(f"Valor p de la pendiente: {p_value_slope:.4e}")

    # 4. Visualización (Histograma y Gráfico de Dispersión con Línea de Regresión)
    plt.figure(figsize=(12, 4))

    # Histograma de calificaciones
    plt.subplot(1, 2, 1)
    sns.histplot(y, kde=True, color='skyblue')
    plt.title(f'Histograma de Calificaciones - {archivo}')
    plt.xlabel('Calificación Final')
    plt.ylabel('Frecuencia')

    # Gráfico de dispersión y línea de regresión
    plt.subplot(1, 2, 2)
    sns.regplot(x=x, y=y, color='darkblue', marker='o',
                line_kws={'label': f'Y = {beta_0:.2f} + {beta_2:.2f}X' if 'beta_2' in locals() else f'Y = {beta_0:.2f} + {beta_1:.2f}X'})
    plt.title(f'Regresión Lineal - {archivo}')
    plt.xlabel('Horas de Estudio')
    plt.ylabel('Calificación Final')
    plt.legend()

    plt.tight_layout()
    plt.show()
```
